```python
import matplotlib.pyplot as plt
import numpy as np

def f1(x):
    return x

def f2(x):
    return 60-x

x1 = np.arange(10,16)
x2 = np.arange(15,21)
x3 = np.arange(20,26)
x4 = np.arange(25,31)

x5 = np.arange(30,36)
x6 = np.arange(35,41)
x7 = np.arange(40,46)
x8 = np.arange(45,51)


x9 = np.arange(15,46)
x10 = np.arange(20,41)
x11 = np.arange(25,36)

plt.plot(x9,np.array([45]*31),color="r",ls="-")
plt.plot(x10,np.array([40]*21),color="r",ls="--")
plt.plot(x11,np.array([35]*11),color="r",ls=":")

plt.plot(x1,f2(x1),color="g",ls="-")
plt.plot(x2,f2(x2),color="g",ls="--")
plt.plot(x3,f2(x3),color="g",ls=":")
#plt.plot(x4,f2(x4),color="g",ls="-.")

plt.plot(x8,f1(x8),color="b",ls="-")
plt.plot(x7,f1(x7),color="b",ls="--")
plt.plot(x6,f1(x6),color="b",ls=":")
#plt.plot(x5,f1(x5),color="b",ls="-.")

plt.xlim(0,60)
plt.ylim(25,60)
plt.xlabel("$log_{10}(\epsilon)$")
plt.ylabel("$log_{10}|\kappa(\epsilon)-1|$")

plt.xticks([], [])
plt.yticks([], [])

plt.text(38, 37, 'Truncation Error: $O(\epsilon)$', fontsize = 8,rotation=52,color="b")
plt.text(11, 37, 'Round Off Error: $O(\epsilon^{-1})$', fontsize = 8,rotation=-52,color="g")
plt.text(20, 46, 'Discretization Error: ($\Delta x, \Delta t < \delta$)', fontsize = 8,color="r")
plt.text(26, 41, '($\Delta x, \Delta t \ll \delta$)', fontsize = 8,color="r")
plt.text(26, 36, '($\Delta x, \Delta t \lll \delta$)', fontsize = 8,color="r")
#plt.show()

ax = plt.subplot(111)

ax.plot(60,0, marker=">", ms=5, color='k', transform=ax.get_xaxis_transform(), clip_on=False)
ax.plot(0, 1, marker="^", ms=5, color='k', transform=ax.get_xaxis_transform(), clip_on=False)
ax.spines[['right', 'top']].set_visible(False)

plt.savefig("KappaTheory.png")
```
